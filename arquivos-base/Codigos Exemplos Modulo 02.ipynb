```python
# Exemplo Prático: Segmentação de Clientes com K-Means
# Este exemplo demonstra como aplicar K-Means em um caso real

# %% [markdown]
# ## Caso de Uso: Segmentação de Clientes de E-commerce
#
# Uma loja online quer segmentar seus clientes para criar campanhas de marketing personalizadas.
# Vamos usar K-Means para identificar diferentes perfis de consumo.

# %%
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
import warnings
warnings.filterwarnings('ignore')

# Configuração visual
plt.style.use('seaborn-v0_8-darkgrid')
np.random.seed(42)

# %% [markdown]
# ## 1. Gerando Dados de Clientes
#
# Vamos simular dados de uma loja com diferentes tipos de clientes.

# %%
# Simular dados de clientes com diferentes perfis
n_customers = 1000

# Perfil 1: Clientes Premium (alta renda, alto gasto, muitas compras)
premium = pd.DataFrame({
    'renda_anual': np.random.normal(150000, 30000, 200),
    'gasto_total': np.random.normal(50000, 10000, 200),
    'num_compras': np.random.normal(100, 20, 200),
    'tempo_cliente_meses': np.random.normal(48, 12, 200),
    'idade': np.random.normal(45, 8, 200)
})

# Perfil 2: Clientes Regulares (média renda, gasto moderado)
regular = pd.DataFrame({
    'renda_anual': np.random.normal(60000, 15000, 400),
    'gasto_total': np.random.normal(15000, 5000, 400),
    'num_compras': np.random.normal(40, 15, 400),
    'tempo_cliente_meses': np.random.normal(24, 10, 400),
    'idade': np.random.normal(35, 10, 400)
})

# Perfil 3: Clientes Ocasionais (baixo engajamento)
ocasional = pd.DataFrame({
    'renda_anual': np.random.normal(40000, 10000, 300),
    'gasto_total': np.random.normal(3000, 1500, 300),
    'num_compras': np.random.normal(10, 5, 300),
    'tempo_cliente_meses': np.random.normal(12, 6, 300),
    'idade': np.random.normal(28, 7, 300)
})

# Perfil 4: Clientes Novos Premium (novos mas com alto potencial)
novo_premium = pd.DataFrame({
    'renda_anual': np.random.normal(120000, 25000, 100),
    'gasto_total': np.random.normal(8000, 2000, 100),
    'num_compras': np.random.normal(15, 5, 100),
    'tempo_cliente_meses': np.random.normal(3, 1, 100),
    'idade': np.random.normal(38, 9, 100)
})

# Combinar todos os perfis
df = pd.concat([premium, regular, ocasional, novo_premium], ignore_index=True)

# Adicionar features derivadas
df['ticket_medio'] = df['gasto_total'] / df['num_compras']
df['gasto_mensal'] = df['gasto_total'] / df['tempo_cliente_meses']
df['compras_por_mes'] = df['num_compras'] / df['tempo_cliente_meses']

# Garantir valores positivos
df = df.abs()

print("📊 Dataset de Clientes Criado!")
print(f"Total de clientes: {len(df)}")
print(f"\nFeatures disponíveis:")
print(df.columns.tolist())
print(f"\nEstatísticas descritivas:")
print(df.describe().round(2))

# %% [markdown]
# ## 2. Análise Exploratória dos Dados

# %%
# Visualizar distribuições
fig, axes = plt.subplots(2, 4, figsize=(16, 8))
axes = axes.ravel()

for idx, col in enumerate(df.columns):
    axes[idx].hist(df[col], bins=30, alpha=0.7, color='skyblue', edgecolor='black')
    axes[idx].set_title(f'{col}')
    axes[idx].set_xlabel('')
    axes[idx].set_ylabel('Frequência')

plt.suptitle('Distribuição das Features dos Clientes', fontsize=16, fontweight='bold')
plt.tight_layout()
plt.show()

# Matriz de correlação
plt.figure(figsize=(10, 8))
correlation_matrix = df.corr()
mask = np.triu(np.ones_like(correlation_matrix), k=1)
sns.heatmap(correlation_matrix, mask=mask, annot=True, cmap='coolwarm',
            center=0, square=True, linewidths=1, cbar_kws={"shrink": .8})
plt.title('Correlação entre Features', fontsize=14, fontweight='bold')
plt.show()

# %% [markdown]
# ## 3. Preparação dos Dados para K-Means

# %%
# Selecionar features relevantes para segmentação
features_segmentacao = ['gasto_total', 'num_compras', 'ticket_medio',
                        'gasto_mensal', 'tempo_cliente_meses']

X = df[features_segmentacao].values

# Normalizar os dados (CRUCIAL para K-Means!)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print("✅ Dados normalizados!")
print(f"Shape dos dados: {X_scaled.shape}")

# Visualizar o efeito da normalização
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

# Antes da normalização
axes[0].boxplot(X, labels=features_segmentacao)
axes[0].set_title('Antes da Normalização')
axes[0].set_xticklabels(features_segmentacao, rotation=45)
axes[0].set_ylabel('Valores')

# Depois da normalização
axes[1].boxplot(X_scaled, labels=features_segmentacao)
axes[1].set_title('Depois da Normalização')
axes[1].set_xticklabels(features_segmentacao, rotation=45)
axes[1].set_ylabel('Valores Normalizados')

plt.tight_layout()
plt.show()

# %% [markdown]
# ## 4. Escolhendo o Número Ideal de Clusters

# %%
# Método do Cotovelo
inertias = []
silhouette_scores = []
K = range(2, 11)

for k in K:
    kmeans = KMeans(n_clusters=k, init='k-means++', n_init=10, random_state=42)
    kmeans.fit(X_scaled)
    inertias.append(kmeans.inertia_)
    silhouette_scores.append(silhouette_score(X_scaled, kmeans.labels_))

# Plotar resultados
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Método do Cotovelo
axes[0].plot(K, inertias, 'bo-', linewidth=2, markersize=8)
axes[0].set_xlabel('Número de Clusters (k)')
axes[0].set_ylabel('Inércia (WCSS)')
axes[0].set_title('Método do Cotovelo')
axes[0].grid(True, alpha=0.3)
axes[0].axvline(x=4, color='red', linestyle='--', alpha=0.7, label='k=4 sugerido')
axes[0].legend()

# Coeficiente de Silhueta
axes[1].plot(K, silhouette_scores, 'go-', linewidth=2, markersize=8)
axes[1].set_xlabel('Número de Clusters (k)')
axes[1].set_ylabel('Coeficiente de Silhueta')
axes[1].set_title('Análise de Silhueta')
axes[1].grid(True, alpha=0.3)
best_k = K[np.argmax(silhouette_scores)]
axes[1].axvline(x=best_k, color='red', linestyle='--', alpha=0.7,
                label=f'k={best_k} (melhor silhueta)')
axes[1].legend()

plt.tight_layout()
plt.show()

print(f"🎯 Número ideal de clusters sugerido: k=4")

# %% [markdown]
# ## 5. Aplicando K-Means com k=4

# %%
# Aplicar K-Means com k=4
k_final = 4
kmeans_final = KMeans(n_clusters=k_final, init='k-means++', n_init=20, random_state=42)
clusters = kmeans_final.fit_predict(X_scaled)

# Adicionar clusters ao DataFrame
df['cluster'] = clusters

print("✅ K-Means aplicado com sucesso!")
print(f"\nDistribuição dos clusters:")
print(df['cluster'].value_counts().sort_index())

# %% [markdown]
# ## 6. Visualizando os Clusters

# %%
# Como temos mais de 2 dimensões, vamos usar PCA para visualizar
from sklearn.decomposition import PCA

# Aplicar PCA para reduzir a 2D
pca = PCA(n_components=2)
X_pca = pca.fit_transform(X_scaled)

# Plotar clusters
plt.figure(figsize=(10, 8))
scatter = plt.scatter(X_pca[:, 0], X_pca[:, 1], c=clusters, cmap='viridis',
                     alpha=0.6, edgecolors='black', linewidth=0.5)

# Plotar centroides transformados
centroids_pca = pca.transform(kmeans_final.cluster_centers_)
plt.scatter(centroids_pca[:, 0], centroids_pca[:, 1],
           c='red', s=300, marker='*', edgecolors='black', linewidth=2)

plt.xlabel(f'PC1 ({pca.explained_variance_ratio_[0]:.1%} da variância)')
plt.ylabel(f'PC2 ({pca.explained_variance_ratio_[1]:.1%} da variância)')
plt.title('Visualização dos Clusters de Clientes (PCA)', fontsize=14, fontweight='bold')
plt.colorbar(scatter, label='Cluster')
plt.grid(True, alpha=0.3)
plt.show()

print(f"Variância total explicada pelo PCA: {sum(pca.explained_variance_ratio_):.1%}")

# %% [markdown]
# ## 7. Analisando o Perfil de Cada Cluster

# %%
# Calcular estatísticas por cluster
cluster_summary = df.groupby('cluster').agg({
    'renda_anual': ['mean', 'count'],
    'gasto_total': 'mean',
    'num_compras': 'mean',
    'ticket_medio': 'mean',
    'tempo_cliente_meses': 'mean',
    'idade': 'mean'
}).round(2)

print("📊 Perfil dos Clusters:")
print(cluster_summary)

# Visualizar perfis em gráfico de radar
from math import pi

# Preparar dados para radar chart
categories = ['Gasto Total', 'Nº Compras', 'Ticket Médio',
              'Tempo Cliente', 'Idade']

fig, axes = plt.subplots(2, 2, figsize=(12, 12), subplot_kw=dict(projection='polar'))
axes = axes.ravel()

for cluster_id in range(k_final):
    # Dados do cluster (normalizados 0-1 para visualização)
    cluster_data = df[df['cluster'] == cluster_id][
        ['gasto_total', 'num_compras', 'ticket_medio',
         'tempo_cliente_meses', 'idade']
    ].mean()

    # Normalizar para escala 0-1
    cluster_data_norm = (cluster_data - df[['gasto_total', 'num_compras',
                                           'ticket_medio', 'tempo_cliente_meses',
                                           'idade']].min()) / \
                       (df[['gasto_total', 'num_compras', 'ticket_medio',
                           'tempo_cliente_meses', 'idade']].max() -
                        df[['gasto_total', 'num_compras', 'ticket_medio',
                           'tempo_cliente_meses', 'idade']].min())

    # Configurar radar
    angles = [n / len(categories) * 2 * pi for n in range(len(categories))]
    values = cluster_data_norm.tolist()
    values += values[:1]  # Fechar o polígono
    angles += angles[:1]

    # Plotar
    ax = axes[cluster_id]
    ax.plot(angles, values, 'o-', linewidth=2)
    ax.fill(angles, values, alpha=0.25)
    ax.set_xticks(angles[:-1])
    ax.set_xticklabels(categories)
    ax.set_ylim(0, 1)
    ax.set_title(f'Cluster {cluster_id}\n({len(df[df["cluster"]==cluster_id])} clientes)',
                 fontsize=12, fontweight='bold')
    ax.grid(True)

plt.suptitle('Perfil de Cada Cluster', fontsize=16, fontweight='bold')
plt.tight_layout()
plt.show()

# %% [markdown]
# ## 8. Interpretação e Nomeação dos Clusters

# %%
# Análise detalhada de cada cluster
cluster_names = {}

for cluster_id in range(k_final):
    cluster_df = df[df['cluster'] == cluster_id]

    print(f"\n{'='*50}")
    print(f"CLUSTER {cluster_id} - {len(cluster_df)} clientes ({len(cluster_df)/len(df)*100:.1f}%)")
    print(f"{'='*50}")

    # Características principais
    avg_gasto = cluster_df['gasto_total'].mean()
    avg_compras = cluster_df['num_compras'].mean()
    avg_tempo = cluster_df['tempo_cliente_meses'].mean()
    avg_ticket = cluster_df['ticket_medio'].mean()

    print(f"💰 Gasto Total Médio: R$ {avg_gasto:,.2f}")
    print(f"🛒 Número de Compras: {avg_compras:.0f}")
    print(f"🎟️  Ticket Médio: R$ {avg_ticket:,.2f}")
    print(f"📅 Tempo como Cliente: {avg_tempo:.0f} meses")

    # Nomear clusters baseado nas características
    if avg_gasto > 40000 and avg_compras > 80:
        nome = "Premium VIP"
        descricao = "Clientes de alto valor, muito engajados"
    elif avg_gasto > 10000 and avg_compras > 30:
        nome = "Clientes Fiéis"
        descricao = "Base sólida de clientes regulares"
    elif avg_tempo < 6 and avg_ticket > 300:
        nome = "Novos Alto Potencial"
        descricao = "Novos clientes com alto ticket médio"
    else:
        nome = "Ocasionais"
        descricao = "Clientes de baixo engajamento"

    cluster_names[cluster_id] = nome
    print(f"\n🏷️  Nome sugerido: {nome}")
    print(f"📝 Descrição: {descricao}")

# %% [markdown]
# ## 9. Estratégias de Marketing por Cluster

# %%
# Criar visualização final com nomes dos clusters
df['cluster_nome'] = df['cluster'].map(cluster_names)

# Box plots comparativos
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

features_plot = ['gasto_total', 'num_compras', 'ticket_medio', 'gasto_mensal']
titles = ['Gasto Total (R$)', 'Número de Compras',
          'Ticket Médio (R$)', 'Gasto Mensal (R$)']

for idx, (feature, title) in enumerate(zip(features_plot, titles)):
    ax = axes[idx // 2, idx % 2]
    df.boxplot(column=feature, by='cluster_nome', ax=ax)
    ax.set_title(title)
    ax.set_xlabel('Segmento de Cliente')
    ax.set_ylabel('')

plt.suptitle('Comparação entre Segmentos de Clientes', fontsize=16, fontweight='bold')
plt.tight_layout()
plt.show()

# Recomendações de marketing
print("\n🎯 ESTRATÉGIAS DE MARKETING POR SEGMENTO:\n")

estrategias = {
    "Premium VIP": [
        "✓ Programa VIP exclusivo com benefícios premium",
        "✓ Acesso antecipado a novos produtos",
        "✓ Eventos exclusivos e experiências personalizadas",
        "✓ Atendimento dedicado e consultoria personalizada"
    ],
    "Clientes Fiéis": [
        "✓ Programa de fidelidade com pontos e recompensas",
        "✓ Descontos progressivos baseados em histórico",
        "✓ Campanhas de cross-selling e up-selling",
        "✓ Newsletter com conteúdo relevante"
    ],
    "Novos Alto Potencial": [
        "✓ Onboarding personalizado",
        "✓ Incentivos para segunda compra",
        "✓ Comunicação educativa sobre produtos",
        "✓ Acompanhamento proativo pós-venda"
    ],
    "Ocasionais": [
        "✓ Campanhas de reativação com ofertas especiais",
        "✓ Email marketing com produtos populares",
        "✓ Simplificar processo de compra",
        "✓ Pesquisas para entender barreiras"
    ]
}

for cluster_id, nome in cluster_names.items():
    print(f"\n{nome.upper()}:")
    for estrategia in estrategias.get(nome, []):
        print(f"  {estrategia}")

# %% [markdown]
# ## 10. Salvando os Resultados

# %%
# Preparar dados para exportação
resultado_final = df[['cluster', 'cluster_nome'] + features_segmentacao].copy()

# Adicionar probabilidade de pertencimento (distância ao centroide)
from scipy.spatial.distance import cdist

distances = cdist(X_scaled, kmeans_final.cluster_centers_, 'euclidean')
probabilities = 1 / (1 + distances)  # Converter distância em "probabilidade"
probabilities = probabilities / probabilities.sum(axis=1, keepdims=True)

# Adicionar probabilidade do cluster atribuído
resultado_final['probabilidade_cluster'] = [
    probabilities[i, cluster] for i, cluster in enumerate(clusters)
]

print("📊 Amostra dos resultados finais:")
print(resultado_final.head(10))

# Estatísticas finais
print(f"\n✅ SEGMENTAÇÃO CONCLUÍDA!")
print(f"   - Total de clientes: {len(df)}")
print(f"   - Número de segmentos: {k_final}")
print(f"   - Coeficiente de Silhueta: {silhouette_score(X_scaled, clusters):.3f}")

# Em produção, você salvaria em arquivo:
# resultado_final.to_csv('segmentacao_clientes.csv', index=False)

# %% [markdown]
# ## Conclusões
#
# ### O que fizemos:
# 1. **Preparamos** dados de clientes com múltiplas features
# 2. **Normalizamos** os dados (essencial para K-Means!)
# 3. **Escolhemos** k=4 usando método do cotovelo e silhueta
# 4. **Aplicamos** K-Means e identificamos 4 segmentos distintos
# 5. **Interpretamos** cada cluster e demos nomes significativos
# 6. **Criamos** estratégias de marketing personalizadas
#
# ### Próximos passos:
# - Validar segmentação com equipe de negócios
# - Implementar campanhas piloto por segmento
# - Monitorar métricas de sucesso
# - Atualizar segmentação periodicamente

print("\n🎉 Exemplo completo de aplicação do K-Means em caso real!")
```
